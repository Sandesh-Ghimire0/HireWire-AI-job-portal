```python
import re
import nltk
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer
```

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords

stop_words = stopwords.words('english')
print(stop_words[:10])
```

```python
stop_words = set(stopwords.words('english'))

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)  # remove special chars
    words = text.split()
    words = [w for w in words if w not in stop_words]
    return " ".join(words)
```

```python
preprocess("hello@ my name is dinesh t@mang")
```

```python
def compute_tfidf_score(resume, job_desc):
    vectorizer = TfidfVectorizer()
    vectors = vectorizer.fit_transform([resume, job_desc])
    
    
    similarity = cosine_similarity(vectors[0], vectors[1])[0][0]
    return similarity * 100
```

```python
job_desc = """7+ years of experience in project management within a software or technology company.

Proven proficiency in managing complex Agile and Scrum projects using Jira and Asana.

Strong understanding of Agile, Scrum, and Kanban methodologies for software delivery.

Excellent leadership, conflict resolution, strategic planning, and problem-solving skills.

Experience managing cross-functional teams delivering SaaS applications.

Familiarity with tools like Trello, Slack, and Microsoft Project for team collaboration."""


resume_desc = """Key Skills

Methodologies: Agile, Scrum, Kanban, Waterfall

Tools: Jira, Asana, Trello, Microsoft Project, Slack

Soft Skills: Strategic Planning, Conflict Resolution, Budgeting, Problem-Solving, Leadership

Work Experience

Senior Project Manager | TechFlow Solutions | 2021 – Present

Managing cross-functional teams of 15+ in delivering customer-facing SaaS applications.

Improved project delivery speed by 30% using Agile and Scrum sprint velocity tracking in Jira.

Led strategic planning sessions and conflict resolution workshops across software teams.

Project Coordinator | CloudData Inc. | 2018 – 2021

Managed Agile project documentation and communication workflows using Asana and Slack.

Coordinated cross-functional teams for large-scale enterprise software deployments."""
```

```python
tfidf_score=compute_tfidf_score(resume=resume_desc,job_desc=job_desc)
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')

def compute_bert_score(resume, job_desc):
    emb1 = model.encode(resume)
    emb2 = model.encode(job_desc)
    
    similarity = cosine_similarity([emb1], [emb2])[0][0]
    return similarity * 100
```

```python
bert_score=compute_bert_score(resume=resume_desc,job_desc=job_desc)
```

```python
def final_score(tfidf_score, bert_score):
    return 0.6 * tfidf_score + 0.4 * bert_score
```

```python
final_score(tfidf_score=tfidf_score,bert_score=bert_score)
```

```python
##end to end pipeline
def analyze_resume(resume_text, job_desc):
    
    resume_clean = preprocess(resume_text)
    job_clean = preprocess(job_desc)
    
    tfidf_score = compute_tfidf_score(resume_clean, job_clean)
    bert_score = compute_bert_score(resume_clean, job_clean)
    
    final_value = final_score(tfidf_score, bert_score)
    
    
    if final_value >= 80:
        confidence = "High"
    elif final_value >= 60:
        confidence = "Medium"
    else:
        confidence = "Low"
    
    return {
        "ATS Score (TF-IDF)": round(tfidf_score, 2),
        "BERT Score": round(bert_score, 2),
        "Final Score": round(final_value, 2),
        "Confidence": confidence
    }
```
